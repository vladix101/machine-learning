```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Python/Position_Salaries.csv')
```

```python
X = df[['Level']].values
y = df['Salary'].values
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators=10, random_state=0)
regressor.fit(X, y)
```

```python
regressor.predict([[6.5]])
```

```python
X_grid = np.arange(X.min(),X.max(),0.01).reshape(-1,1)
plt.scatter(X,y)
plt.plot(X_grid,regressor.predict(X_grid),'r')
```
